```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

df = pd.read_csv("../data/customer_support_tickets.csv")
df.head()
```

```python
df.shape
df.info()
```

```python
df.isnull().sum()
```

```python
print(df.duplicated().sum())
print(df["language"].value_counts())
```

```python
df["queue"].value_counts()
```

```python
df["queue"].value_counts().plot(kind="bar", figsize=(10,5))
plt.title("Distribution of Support Queues")
plt.xlabel("Queue")
plt.ylabel("Number of Tickets")
plt.xticks(rotation=45)
plt.show()
```

```python
df["priority"].value_counts()
```

```python
df["priority"].value_counts().plot(kind="bar")
plt.title("Distribution of Ticket Priorities")
plt.xlabel("Priority")
plt.ylabel("Number of Tickets")
plt.show()
```

```python
df["type"].value_counts()
```

```python
df["type"].value_counts().plot(kind="bar")
plt.title("Distribution of Ticket Types")
plt.xlabel("Ticket Type")
plt.ylabel("Number of Tickets")
plt.show()
```

```python
df["body_length"] = df["body"].str.split().str.len()
df["body_length"].describe()
```

```python
df["body_length"].hist(bins=30, figsize=(8,5))

plt.title("Distribution of Ticket Length")
plt.xlabel("Number of Words")
plt.ylabel("Frequency")

plt.show()
```
