```python
import pandas as pd

df = pd.read_csv("../data/customer_support_tickets_clean.csv")
sample_ticket = df.loc[0, "ticket_text"]

print(sample_ticket)
```

```python
prompt = f"""
You are a professional customer support representative.

Read the customer ticket below and write ONLY the customer support reply.

Requirements:
- Begin by thanking the customer.
- Apologize for the inconvenience.
- Address the customer's issue.
- Assure them the issue is being investigated.
- Keep the response professional and polite.
- Do NOT repeat or copy the customer's ticket.
- Do NOT include the original ticket in your answer.

Customer Ticket:
{sample_ticket}

Customer Support Reply:
"""

response = generator(
    prompt,
    max_new_tokens=120,
    do_sample=False
)

print(response[0]["generated_text"])
```

```python
from transformers import pipeline

generator = pipeline(
    "text-generation",
    model="HuggingFaceTB/SmolLM2-360M-Instruct"
)
```

```python
messages = [
    {
        "role": "system",
        "content": """
You are an AI Customer Support Copilot.

Generate ONLY the body of a professional customer support reply.

Your response should:
- Acknowledge the customer's concern.
- Apologize sincerely for the inconvenience.
- Show empathy in a professional manner.
- Inform the customer that the support team is actively investigating the issue.
- Assure the customer that resolving the issue is a priority.
- Thank the customer for their patience and understanding.

Do NOT:
- Include greetings such as "Dear Customer", "Hello", or "Hi".
- Include any closing or signature.
- Repeat or summarize the customer's issue.
- Ask for additional information unless absolutely necessary.
- Mention placeholders like [Customer Name].
- Invent timelines or technical details.

Write naturally as if the message will be sent directly to the customer.

Return ONLY the reply body.
"""
    },
    {
        "role": "user",
        "content": f"""
Customer Issue:

{summary}

Write the reply body.
"""
    }
]

response = generator(
    messages,
    max_new_tokens=100,
    do_sample=False
)

body = response[0]["generated_text"][-1]["content"]

final_reply = f"""Dear Customer,

{body}

Kind regards,
Customer Support Team"""

print(final_reply)
```

```python
sample_df = df.sample(10, random_state=42)
```

```python
for i, row in sample_df.iterrows():

    ticket = row["ticket_text"]

    summary = ticket

    messages = [
        {
            "role": "system",
            "content": """
You are an AI Customer Support Copilot.

Generate ONLY the body of a professional customer support reply.

Do not include greetings or signatures.
Reply directly to the customer.
Apologize for the inconvenience.
Acknowledge the issue.
Reassure the customer that the issue is being investigated.
Thank them for their patience.
Return only the reply body.
"""
        },
        {
            "role": "user",
            "content": f"Customer Issue:\n\n{summary}"
        }
    ]

    response = generator(
        messages,
        max_new_tokens=100,
        do_sample=False
    )

    body = response[0]["generated_text"][-1]["content"]

    print("=" * 100)
    print("ORIGINAL:\n")
    print(ticket)

    print("\nSUMMARY:\n")
    print(summary)

    print("\nREPLY:\n")
    print(body)

    print("=" * 100)
```

```python
messages = [
    {
        "role": "system",
        "content": """
You are a professional Customer Support Agent.

Your task is to write a polite, professional, and empathetic email reply to the customer.

Rules:
- Start with "Dear Customer,"
- Thank the customer for contacting support.
- Apologize for the inconvenience caused.
- Acknowledge the customer's issue.
- Reassure the customer that the support team is investigating the issue.
- If the customer has requested information, politely mention that the request will be reviewed and an update will be provided.
- Do NOT invent any facts.
- Do NOT assume the cause, duration, or resolution of the issue.
- Do NOT mention information that is not present in the customer's message.
- Do NOT ask unnecessary questions unless absolutely required.
- Do NOT use placeholders like [Customer Name].
- Keep the reply concise (5–8 sentences).
- End exactly with:

Kind regards,
Customer Support Team

Return ONLY the final reply.
"""
    },
    {
        "role": "user",
        "content": sample_ticket
    }
]

response = generator(
    messages,
    max_new_tokens=220,
    do_sample=False
)

print(response[0]["generated_text"][-1]["content"])


response = generator(
    messages,
    max_new_tokens=220,
    do_sample=False
)
```
